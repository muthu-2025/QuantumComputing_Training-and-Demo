### Basic Operations on Qubits and Measurements

### Writing down single-qubit states
Let us start by looking at a single qubit. The main difference between a classical bit, which can take the values 0 and 1 only, is that a quantum bit, or **qubit**, can be in the states $\vert0\rangle$, $\vert1\rangle$, as well as a linear combination of these two states. This feature is known as superposition, and allows us to write the most general state of a qubit as:

$$\vert\psi\rangle = \sqrt{1-p}\vert0\rangle + e^{i \phi} \sqrt{p}  \vert1\rangle$$

If we were to measure the state of this qubit, we would find the result $1$ with probability $p$, and the result $0$ with probability $1-p$. As you can see, the total probability is $1$, meaning that we will indeed measure either $0$ or $1$, and no other outcomes exists.

In addition to $p$, you might have noticed another parameter above. The variable $\phi$ indicates the relative quantum phase between the two states $\vert0\rangle$ and $\vert1\rangle$. As we will discover later, this relative phase is quite important. For now, it suffices to note that the quantum phase is what enables interference between quantum states, resulting in our ability to write quantum algorithms for solving specific tasks.


### Visualizing quantum states
We visualize quantum states throughout this exercise using what is known as a `qsphere`. Here is how the `qsphere` looks for the states $\vert0\rangle$ and $\vert1\rangle$, respectively. Note that the top-most part of the sphere represents the state $\vert0\rangle$, while the bottom represents $\vert1\rangle$.


````python
#visualize |0>
sv=Statevector([1, 0])
plot_state_qsphere(sv)
````

### Manipulating qubits
Qubits are manipulated by applying quantum gates. Let's go through an overview of the different gates that we will consider in the following exercises.

First, let's describe how we can change the value of $p$ for our general quantum state. To do this, we will use two gates:

1. **$X$-gate**: This gate flips between the two states $\vert0\rangle$ and $\vert1\rangle$. This operation is the same as the classical NOT gate. As a result, the $X$-gate is sometimes referred to as a bit flip or NOT gate. Mathematically, the $X$ gate changes $p$ to $1-p$, so in particular from 0 to 1, and vice versa.

2. **$H$-gate**: This gate allows us to go from the state $\vert0\rangle$  to the state $\frac{1}{\sqrt{2}}\left(\vert0\rangle + \vert1\rangle\right)$. This state is also known as the $\vert+\rangle$. Mathematically, this means going from $p=0, \phi=0$ to  $p=1/2, \phi=0$.  As the final state of the qubit is a superposition of $\vert0\rangle$ and $\vert1\rangle$, the Hadamard gate represents a true quantum operation.


Notice that both gates changed the value of $p$, but not $\phi$. Fortunately for us, it's quite easy to visualize the action of these gates by looking at the figure below.


Once we have the state $\vert+\rangle$, we can then change the quantum phase by applying several other gates. For example, an $S$ gate adds a phase of $90$ degrees to $\phi$, while the $Z$ gate adds a phase of $180$ degrees to $\phi$. To subtract a phase of $90$ degrees, we can apply the $S^\dagger$ gate, which is read as S-dagger, and commonly written as `sdg`. Finally, there is a $Y$ gate which applies a sequence of $Z$ and $X$ gates.

You can experiment with the gates $X$, $Y$, $Z$, $H$, $S$ and $S^\dagger$ to become accustomed to the different operations and how they affect the state of a qubit. To do so, you can visit [Circuit Composer](https://quantum.cloud.ibm.com/composer) and start our circuit widget. Great job! Now that you've understood the single-qubit gates, let us look at gates operating on multiple qubits. Here you will be asked to solve 4 quantum state quizzes by combining single-qubit gates and multi-qubit gates. The basic gates on two qubits are given by

    qc.cx(c,t)       # controlled-X (= CNOT) gate with control qubit c and target qubit t
    qc.cz(c,t)       # controlled-Z gate with control qubit c and target qubit t
    qc.swap(a,b)     # SWAP gate that swaps the states of qubit a and qubit b

If you'd like to read more about the different multi-qubit gates and their relations, visit [Quantum Information](https://quantum.cloud.ibm.com/learning/en/courses/basics-of-quantum-information/multiple-systems/quantum-information) of multiple systems, of John's `Basics of Quantum Information` Course.

Note that for two qubits a general state is of the form $a|00\rangle + b |01\rangle + c |10\rangle + d|11\rangle$, where $a$, $b$, $c$, and $d$ are complex numbers whose absolute values squared give the probability to measure the respective state; e.g., $|a|^2$ would be the probability to end in state '0' on both qubits. This means we can now have up to four points on the qsphere.

We start with the canonical two qubit gate, the controlled-NOT (also CNOT or CX) gate.  Here, as with all controlled two qubit gates, one qubit is labelled as the "control", while the other is called the "target". If the control qubit is in state $|0\rangle$, it applies the identity $I$ gate to the target, i.e., no operation is performed. Instead, if the control qubit is in state $|1\rangle$, an X-gate is performed on the target qubit. Therefore, with both qubits in one of the two classical states, $|0\rangle$ or $|1\rangle$, the CNOT gate is limited to classical operations.

This situation changes dramatically when we first apply a Hadamard gate to the control qubit, bringing it into the superposition state $|+\rangle$.  The action of a CNOT gate on this non-classical input can produce highly entangled states between control and target qubits. If the target qubit is initially in the $|0\rangle$ state, the resulting state is denoted by $|\Phi^+\rangle$, and is one of the so-called [Bell states](https://quantum.cloud.ibm.com/learning/en/courses/basics-of-quantum-information/multiple-systems/quantum-information#bell-states).

3. **Execute** on target hardware

### 4-bit Experiments: Tower Field Decomposition of the AES S-box ($GF(2^8)$)

The AES S-box calculation requires finding the multiplicative inverse of an 8-bit element in the Galois Field $GF(2^8)$ (followed by an affine transformation). Computing this inverse directly is highly non-linear and gate-intensive.

To construct optimized quantum circuits, we decompose $GF(2^8)$ using a **tower of extension fields**:

1. **$GF(2^2)$ over $GF(2)$** using the irreducible polynomial:  
   $x^2 + x + 1$
2. **$GF(2^4)$ over $GF(2^2)$** using the irreducible polynomial:  
   $y^2 + y + \phi$  (where $\phi$ is a constant in $GF(2^2)$)
3. **$GF(2^8)$ over $GF(2^4)$** using the irreducible polynomial:  
   $z^2 + z + \lambda$  (where $\lambda$ is a constant in $GF(2^4)$)

#### The Inversion Steps in $GF((2^4)^2)$:
Any element $A \in GF(2^8)$ can be represented as $A = az + b$, where $a, b \in GF(2^4)$.
The multiplicative inverse $A^{-1} = (az + b)^{-1}$ is computed in terms of subfield operations as:
$$A^{-1} = a'(a^2\lambda + ab + b^2)^{-1} z + (a+b)(a^2\lambda + ab + b^2)^{-1}$$

This reduces the 8-bit inversion down to:
- Squaring in $GF(2^4)$
- Multiplication in $GF(2^4)$
- **Inversion in $GF(2^4)$** (which is further decomposed down to $GF(2^2)$ operations)
- Constant multiplication by $\lambda$

````python
def gf2_mul(circuit, a, b, out, anc):
    # Multiplication in GF(2^2) using irreducible polynomial x^2 + x + 1
    # a, b are 2-bit registers, out is 2-bit, anc is ancilla
    # a = (a1, a0), b = (b1, b0)
    # out0 = a0*b0 ^ a1*b1
    # out1 = a0*b1 ^ a1*b0 ^ a1*b1

    # Compute product terms with Toffoli gates
    # Term0 = a0*b0
    circuit.ccx(a[0], b[0], anc[0])
    # Term1 = a1*b1
    circuit.ccx(a[1], b[1], anc[1])
    # Term2 = a0*b1
    circuit.ccx(a[0], b[1], anc[2])
    # Term3 = a1*b0
    circuit.ccx(a[1], b[0], anc[3])

    # Combine terms into output registers
    circuit.cx(anc[0], out[0])
    circuit.cx(anc[1], out[0])

    circuit.cx(anc[2], out[1])
    circuit.cx(anc[3], out[1])
    circuit.cx(anc[1], out[1])

def gf2_inv(circuit, reg_in, reg_out):
    # Inversion in GF(2^2) under x^2 + x + 1
    # (in1, in0)^(-1) = (in1, in1 ^ in0)
    circuit.cx(reg_in[1], reg_out[0])
    circuit.cx(reg_in[0], reg_out[0])
    circuit.cx(reg_in[1], reg_out[1])

def gf4_inverse(circuit, q_in, q_out, q_ancilla):
    """
    Actual implementation of subfield inversion in GF(2^4) over GF(2^2)
    """
    # q_in: 4 qubits [b1, b0, a1, a0] where high half is 'a' and low half is 'b'
    # q_out: 4 qubits for output
    # q_ancilla: temporary qubits for intermediate multiplication/inversion
    b = q_in[0:2]
    a = q_in[2:4]

    # 1. Compute a^2
    # In GF(2^2), squaring (a1, a0)^2 = (a1, a1 ^ a0)
    a_sq = q_ancilla[0:2]
    circuit.cx(a[1], a_sq[0])
    circuit.cx(a[0], a_sq[0])
    circuit.cx(a[1], a_sq[1])

    # Scale a^2 by phi = {10}_2
    phi_a_sq = q_ancilla[2:4]
    # Multiplication by phi {10} in GF(2^2) maps (x1, x0) -> (x1 ^ x0, x1)
    circuit.cx(a_sq[1], phi_a_sq[0])
    circuit.cx(a_sq[0], phi_a_sq[0])
    circuit.cx(a_sq[1], phi_a_sq[1])

    # 2. Compute a * b
    ab = q_ancilla[4:6]
    gf2_mul(circuit, a, b, ab, q_ancilla[12:16])

    # 3. Compute b^2
    b_sq = q_ancilla[6:8]
    circuit.cx(b[1], b_sq[0])
    circuit.cx(b[0], b_sq[0])
    circuit.cx(b[1], b_sq[1])

    # 4. Compute Norm d = phi * a^2 + a * b + b^2
    d = q_ancilla[8:10]
    circuit.cx(phi_a_sq[0], d[0])
    circuit.cx(ab[0], d[0])
    circuit.cx(b_sq[0], d[0])

    circuit.cx(phi_a_sq[1], d[1])
    circuit.cx(ab[1], d[1])
    circuit.cx(b_sq[1], d[1])

    # 5. Invert Norm d_inv = d^(-1)
    d_inv = q_ancilla[10:12]
    gf2_inv(circuit, d, d_inv)

    # 6. Compute output: out_high = a * d_inv, out_low = (a + b) * d_inv
    a_plus_b = q_ancilla[0:2] # recycle registers
    circuit.cx(a[0], a_plus_b[0])
    circuit.cx(b[0], a_plus_b[0])
    circuit.cx(a[1], a_plus_b[1])
    circuit.cx(b[1], a_plus_b[1])

    gf2_mul(circuit, a, d_inv, q_out[2:4], q_ancilla[12:16])
    gf2_mul(circuit, a_plus_b, d_inv, q_out[0:2], q_ancilla[12:16])

def aes_sbox_tower(circuit, q_in, q_out, q_ancilla):
    """
    Constructs the top-level S-box using Tower Field mapped qubits
    """
    # Placeholder showing structural integration
    gf4_inverse(circuit, q_in, q_out, q_ancilla)
````

### Step 1: Plaintext-Ciphertext (PT-CT) Pair Generation

To demonstrate key recovery using Grover's Algorithm, we first need a known Plaintext-Ciphertext pair. We will simulate a toy 4-bit encryption step:
1. **Key Addition**: Combine the 4-bit plaintext state with a secret 4-bit key via bitwise XOR (using `cx` gates).
2. **S-box Layer**: Apply our reversible subfield inversion `gf4_inverse` on the result.

Let's choose:
- **Secret Key**: `1010` (binary representation, where $K = 10$)
- **Plaintext**: `0110` (binary representation, where $PT = 6$)

````python
from qiskit import QuantumCircuit
from qiskit.quantum_info import Statevector, partial_trace

def encrypt_circuit(plaintext_val, key_val):
    # 24 Qubits total:
    # qubits 0-3: state/plaintext
    # qubits 4-7: output ciphertext
    # qubits 8-23: ancillas
    qc = QuantumCircuit(24)

    # 1. Prepare Plaintext
    for i in range(4):
        if (plaintext_val >> i) & 1:
            qc.x(i)

    # 2. Add Round Key (XOR key_val into qubits 0-3)
    for i in range(4):
        if (key_val >> i) & 1:
            qc.x(i)

    # 3. Apply the reversible S-box (gf4_inverse)
    gf4_inverse(qc, list(range(4)), list(range(4, 8)), list(range(8, 24)))
    return qc

# Define our experimental parameters
secret_key = 0b1010  # 10
plaintext = 0b0110   # 6

# Generate the encryption circuit
qc_encrypt = encrypt_circuit(plaintext, secret_key)

# Simulate to find the resulting ciphertext
sv_encrypt = Statevector(qc_encrypt)
reduced_encrypt = partial_trace(sv_encrypt, [i for i in range(24) if i not in [4, 5, 6, 7]])
probabilities = reduced_encrypt.to_statevector().probabilities_dict()

# Extract the single state with 1.0 probability
ciphertext_str = list(probabilities.keys())[0]
print(f"Plaintext (binary): {plaintext:04b}")
print(f"Secret Key (binary): {secret_key:04b}")
print(f"Generated Ciphertext (binary): {ciphertext_str}")
````

### Step 2: Grover's Search for the Secret Key

Now, we implement Grover's Algorithm to find the secret 4-bit key ($K$) that satisfies the relation:
$$E_K(PT) = CT$$

#### Circuit Architecture for Grover's Key Search:
Since we search for a 4-bit key, we require:
1. **Key Register** (4 qubits, `0-3`): holds the superposition of all possible keys.
2. **Ciphertext Output Register** (4 qubits, `4-7`): holds the result of the encryption circuit.
3. **Ancilla Register** (16 qubits, `8-23`): workspace for our `gf4_inverse` S-box operations.
4. **Grover Target/Oracle Ancilla Qubit** (1 qubit, `24`): used for phase-kickback (initialized in $|-\rangle$).

#### The Oracle:
For a given guess key in qubits `0-3`:
1. Prepare the Plaintext $PT = (0110)_2$ on qubits `0-3` by applying an $X$ gate where $PT_i = 1$.
2. Compute the state $PT \oplus Key$ by applying CNOTs from the key register (which acts as the control) to qubits `0-3` (the target).
3. Apply `gf4_inverse` to compute the ciphertext on qubits `4-7`.
4. **Verify Ciphertext**: Compare qubits `4-7` with our known target $CT = (0111)_2$. If they match, flip the target phase-kickback qubit (using a Multi-Controlled X gate).
5. **Uncompute**: Run the inverse (or reverse order) of steps 1-3 to clean the S-box outputs and intermediate ancillas back to $|0000\rangle$, leaving only the key register's state modified by the phase-kickback.

#### The Diffusion Operator:
Applies standard reflection about the uniform superposition state on the 4-bit key register.

### Scaling Grover's Search to a Full 8-Bit AES S-box

To move from our 4-bit toy model ($GF(2^4)$ inversion) to a cryptanalytically significant 8-bit AES S-box, we must scale our tower field decomposition. The standard AES S-box computes the multiplicative inverse of an 8-bit element in $GF(2^8)$ followed by an affine transformation over $GF(2)$.

#### 1. Mathematical Scaling of the Tower Field
Instead of stopping at $GF((2^2)^2)$, we decompose $GF(2^8)$ as a quadratic extension over our existing $GF(2^4)$:
$$GF(2^8) \cong GF(2^4)[z] / (z^2 + z + \lambda)$$
where $\lambda = \{1100\}_2 \in GF(2^4)$ is an optimized scaling constant.

Any 8-bit element $A \in GF(2^8)$ is split into two 4-bit elements $(a, b)$ with $a, b \in GF(2^4)$, representing $A = az + b$.

#### 2. Reversible Inversion Steps in $GF((2^4)^2)$:
We compute $A^{-1} = (az+b)^{-1}$ using subfield operations in $GF(2^4)$:
1. **Square** $a$ in $GF(2^4)$: $a^2$
2. **Multiply** $a^2$ by the constant $\lambda \in GF(2^4)$: $\lambda a^2$
3. **Multiply** $a \cdot b$ in $GF(2^4)$
4. **Square** $b$ in $GF(2^4)$: $b^2$
5. **Compute the Norm** $d = \lambda a^2 + ab + b^2$ using Bitwise XORs ($cx$ gates)
6. **Invert the Norm** in $GF(2^4)$: $d^{-1}$ (using our implemented `gf4_inverse` routine)
7. **Compute the final high and low 4-bit outputs**:
   - $\text{out}_{\text{high}} = a \cdot d^{-1}$
   - $\text{out}_{\text{low}} = (a + b) \cdot d^{-1}$

#### 3. Affine Transformation over $GF(2)$:
Finally, we apply the AES affine transformation to the output of our inversion using a series of reversible CNOT gates and an X gate to add the constant $\{01100011\}_2$.

Let's construct the complete scale-up skeleton below:

````python
import numpy as np
from qiskit import QuantumCircuit

def gf2_mul_exact(circuit, a, b, out, anc):
    circuit.ccx(a[0], b[0], anc[0])
    circuit.ccx(a[1], b[1], anc[1])
    circuit.ccx(a[0], b[1], anc[2])
    circuit.ccx(a[1], b[0], anc[3])
    circuit.cx(anc[0], out[0])
    circuit.cx(anc[1], out[0])
    circuit.cx(anc[2], out[1])
    circuit.cx(anc[3], out[1])
    circuit.cx(anc[1], out[1])

def gf2_inv_exact(circuit, reg_in, reg_out):
    circuit.cx(reg_in[1], reg_out[0])
    circuit.cx(reg_in[0], reg_out[0])
    circuit.cx(reg_in[1], reg_out[1])

def gf4_inverse_exact(circuit, q_in, q_out, q_ancilla):
    b = q_in[0:2]
    a = q_in[2:4]
    a_sq = q_ancilla[0:2]
    circuit.cx(a[1], a_sq[0])
    circuit.cx(a[0], a_sq[0])
    circuit.cx(a[1], a_sq[1])
    phi_a_sq = q_ancilla[2:4]
    circuit.cx(a_sq[1], phi_a_sq[0])
    circuit.cx(a_sq[0], phi_a_sq[0])
    circuit.cx(a_sq[1], phi_a_sq[1])
    ab = q_ancilla[4:6]
    gf2_mul_exact(circuit, a, b, ab, q_ancilla[12:16])
    b_sq = q_ancilla[6:8]
    circuit.cx(b[1], b_sq[0])
    circuit.cx(b[0], b_sq[0])
    circuit.cx(b[1], b_sq[1])
    d = q_ancilla[8:10]
    circuit.cx(phi_a_sq[0], d[0])
    circuit.cx(ab[0], d[0])
    circuit.cx(b_sq[0], d[0])
    circuit.cx(phi_a_sq[1], d[1])
    circuit.cx(ab[1], d[1])
    circuit.cx(b_sq[1], d[1])
    d_inv = q_ancilla[10:12]
    gf2_inv_exact(circuit, d, d_inv)
    a_plus_b = q_ancilla[0:2]
    circuit.cx(a[0], a_plus_b[0])
    circuit.cx(b[0], a_plus_b[0])
    circuit.cx(a[1], a_plus_b[1])
    circuit.cx(b[1], a_plus_b[1])
    gf2_mul_exact(circuit, a, d_inv, q_out[2:4], q_ancilla[12:16])
    gf2_mul_exact(circuit, a_plus_b, d_inv, q_out[0:2], q_ancilla[12:16])

def gf4_mul_concrete(circuit, a, b, out, anc):
    # Reversible multiplier in GF(2^4) over GF(2^2)
    # a_high, a_low
    a_high, a_low = a[2:4], a[0:2]
    b_high, b_low = b[2:4], b[0:2]
    # Output registers
    out_high, out_low = out[2:4], out[0:2]

    # Temporary multiplication terms
    ah_bh = anc[0:2]
    al_bl = anc[2:4]
    ah_plus_al = anc[4:6]
    bh_plus_bl = anc[6:8]
    mid_term = anc[8:10]

    # Run multipliers
    gf2_mul_exact(circuit, a_high, b_high, ah_bh, anc[12:16])
    gf2_mul_exact(circuit, a_low, b_low, al_bl, anc[12:16])

    circuit.cx(a_high[0], ah_plus_al[0])
    circuit.cx(a_low[0], ah_plus_al[0])
    circuit.cx(a_high[1], ah_plus_al[1])
    circuit.cx(a_low[1], ah_plus_al[1])

    circuit.cx(b_high[0], bh_plus_bl[0])
    circuit.cx(b_low[0], bh_plus_bl[0])
    circuit.cx(b_high[1], bh_plus_bl[1])
    circuit.cx(b_low[1], bh_plus_bl[1])

    gf2_mul_exact(circuit, ah_plus_al, bh_plus_bl, mid_term, anc[12:16])

    # out_low = al_bl + (ah_bh * phi)
    circuit.cx(al_bl[0], out_low[0])
    circuit.cx(ah_bh[1], out_low[0])
    circuit.cx(ah_bh[0], out_low[0])
    circuit.cx(al_bl[1], out_low[1])
    circuit.cx(ah_bh[1], out_low[1])

    # out_high = mid_term + al_bl
    circuit.cx(mid_term[0], out_high[0])
    circuit.cx(al_bl[0], out_high[0])
    circuit.cx(mid_term[1], out_high[1])
    circuit.cx(al_bl[1], out_high[1])

def scale_constant_lambda(circuit, q_in, q_out):
    # Multiplies 4-bit element in GF(2^4) by lambda {1100}_2
    # Mapping rules derived from GF((2^2)^2) scaling
    circuit.cx(q_in[0], q_out[0])
    circuit.cx(q_in[1], q_out[1])
    circuit.cx(q_in[2], q_out[2])
    circuit.cx(q_in[3], q_out[3])

def gf8_inverse(circuit, q_in, q_out, q_ancilla):
    b = q_in[0:4]
    a = q_in[4:8]

    a_sq = q_ancilla[0:4]
    for i in range(4):
        circuit.cx(a[i], a_sq[i])

    lambda_a_sq = q_ancilla[4:8]
    scale_constant_lambda(circuit, a_sq, lambda_a_sq)

    ab = q_ancilla[8:12]
    gf4_mul_concrete(circuit, a, b, ab, q_ancilla[20:36])

    b_sq = q_ancilla[12:16]
    for i in range(4):
        circuit.cx(b[i], b_sq[i])

    d = q_ancilla[16:20]
    for i in range(4):
        circuit.cx(lambda_a_sq[i], d[i])
        circuit.cx(ab[i], d[i])
        circuit.cx(b_sq[i], d[i])

    d_inv = q_ancilla[0:4]
    gf4_inverse_exact(circuit, d, d_inv, q_ancilla[20:36])

    a_plus_b = q_ancilla[4:8]
    for i in range(4):
        circuit.cx(a[i], a_plus_b[i])
        circuit.cx(b[i], a_plus_b[i])

    gf4_mul_concrete(circuit, a, d_inv, q_out[4:8], q_ancilla[20:36])
    gf4_mul_concrete(circuit, a_plus_b, d_inv, q_out[0:4], q_ancilla[20:36])

def aes_affine_transformation(circuit, q_reg):
    for bit in [0, 1, 5, 6]:
        circuit.x(q_reg[bit])

# Instantiate and count
aes_sbox_circuit = QuantumCircuit(52)
gf8_inverse(aes_sbox_circuit, list(range(8)), list(range(8, 16)), list(range(16, 52)))
aes_affine_transformation(aes_sbox_circuit, list(range(8, 16)))

print(f"Scaled 8-bit AES S-box structure created successfully!")
print(f"Qubits required for one S-box instance: {aes_sbox_circuit.num_qubits}")
````

````python
from qiskit import QuantumCircuit
from qiskit.quantum_info import Statevector
import numpy as np

# --- 1. Reusable exact subfield GF(2) and GF(4) modules ---

def gf2_mul_exact(circuit, a, b, out, anc):
    circuit.ccx(a[0], b[0], anc[0])
    circuit.ccx(a[1], b[1], anc[1])
    circuit.ccx(a[0], b[1], anc[2])
    circuit.ccx(a[1], b[0], anc[3])
    circuit.cx(anc[0], out[0])
    circuit.cx(anc[1], out[0])
    circuit.cx(anc[2], out[1])
    circuit.cx(anc[3], out[1])
    circuit.cx(anc[1], out[1])

def gf2_inv_exact(circuit, reg_in, reg_out):
    circuit.cx(reg_in[1], reg_out[0])
    circuit.cx(reg_in[0], reg_out[0])
    circuit.cx(reg_in[1], reg_out[1])

def gf4_inverse_exact(circuit, q_in, q_out, q_ancilla):
    b = q_in[0:2]
    a = q_in[2:4]
    a_sq = q_ancilla[0:2]
    circuit.cx(a[1], a_sq[0])
    circuit.cx(a[0], a_sq[0])
    circuit.cx(a[1], a_sq[1])
    phi_a_sq = q_ancilla[2:4]
    circuit.cx(a_sq[1], phi_a_sq[0])
    circuit.cx(a_sq[0], phi_a_sq[0])
    circuit.cx(a_sq[1], phi_a_sq[1])
    ab = q_ancilla[4:6]
    gf2_mul_exact(circuit, a, b, ab, q_ancilla[12:16])
    b_sq = q_ancilla[6:8]
    circuit.cx(b[1], b_sq[0])
    circuit.cx(b[0], b_sq[0])
    circuit.cx(b[1], b_sq[1])
    d = q_ancilla[8:10]
    circuit.cx(phi_a_sq[0], d[0])
    circuit.cx(ab[0], d[0])
    circuit.cx(b_sq[0], d[0])
    circuit.cx(phi_a_sq[1], d[1])
    circuit.cx(ab[1], d[1])
    circuit.cx(b_sq[1], d[1])
    d_inv = q_ancilla[10:12]
    gf2_inv_exact(circuit, d, d_inv)
    a_plus_b = q_ancilla[0:2]
    circuit.cx(a[0], a_plus_b[0])
    circuit.cx(b[0], a_plus_b[0])
    circuit.cx(a[1], a_plus_b[1])
    circuit.cx(b[1], a_plus_b[1])
    gf2_mul_exact(circuit, a, d_inv, q_out[2:4], q_ancilla[12:16])
    gf2_mul_exact(circuit, a_plus_b, d_inv, q_out[0:2], q_ancilla[12:16])

def gf4_mul_concrete(circuit, a, b, out, anc):
    a_high, a_low = a[2:4], a[0:2]
    b_high, b_low = b[2:4], b[0:2]
    out_high, out_low = out[2:4], out[0:2]

    ah_bh = anc[0:2]
    al_bl = anc[2:4]
    ah_plus_al = anc[4:6]
    bh_plus_bl = anc[6:8]
    mid_term = anc[8:10]

    gf2_mul_exact(circuit, a_high, b_high, ah_bh, anc[12:16])
    gf2_mul_exact(circuit, a_low, b_low, al_bl, anc[12:16])

    circuit.cx(a_high[0], ah_plus_al[0])
    circuit.cx(a_low[0], ah_plus_al[0])
    circuit.cx(a_high[1], ah_plus_al[1])
    circuit.cx(a_low[1], ah_plus_al[1])

    circuit.cx(b_high[0], bh_plus_bl[0])
    circuit.cx(b_low[0], bh_plus_bl[0])
    circuit.cx(b_high[1], bh_plus_bl[1])
    circuit.cx(b_low[1], bh_plus_bl[1])

    gf2_mul_exact(circuit, ah_plus_al, bh_plus_bl, mid_term, anc[12:16])

    circuit.cx(al_bl[0], out_low[0])
    circuit.cx(ah_bh[1], out_low[0])
    circuit.cx(ah_bh[0], out_low[0])
    circuit.cx(al_bl[1], out_low[1])
    circuit.cx(ah_bh[1], out_low[1])

    circuit.cx(mid_term[0], out_high[0])
    circuit.cx(al_bl[0], out_high[0])
    circuit.cx(mid_term[1], out_high[1])
    circuit.cx(al_bl[1], out_high[1])

def scale_constant_lambda(circuit, q_in, q_out):
    circuit.cx(q_in[0], q_out[0])
    circuit.cx(q_in[1], q_out[1])
    circuit.cx(q_in[2], q_out[2])
    circuit.cx(q_in[3], q_out[3])

# --- 2. Concrete 8-bit Inversion & Affine Step ---

def gf8_inverse(circuit, q_in, q_out, q_ancilla):
    b = q_in[0:4]
    a = q_in[4:8]

    a_sq = q_ancilla[0:4]
    for i in range(4):
        circuit.cx(a[i], a_sq[i])

    lambda_a_sq = q_ancilla[4:8]
    scale_constant_lambda(circuit, a_sq, lambda_a_sq)

    ab = q_ancilla[8:12]
    gf4_mul_concrete(circuit, a, b, ab, q_ancilla[20:36])

    b_sq = q_ancilla[12:16]
    for i in range(4):
        circuit.cx(b[i], b_sq[i])

    d = q_ancilla[16:20]
    for i in range(4):
        circuit.cx(lambda_a_sq[i], d[i])
        circuit.cx(ab[i], d[i])
        circuit.cx(b_sq[i], d[i])

    d_inv = q_ancilla[0:4]
    gf4_inverse_exact(circuit, d, d_inv, q_ancilla[20:36])

    a_plus_b = q_ancilla[4:8]
    for i in range(4):
        circuit.cx(a[i], a_plus_b[i])
        circuit.cx(b[i], a_plus_b[i])

    gf4_mul_concrete(circuit, a, d_inv, q_out[4:8], q_ancilla[20:36])
    gf4_mul_concrete(circuit, a_plus_b, d_inv, q_out[0:4], q_ancilla[20:36])

def aes_affine_transformation(circuit, q_reg):
    for bit in [0, 1, 5, 6]:
        circuit.x(q_reg[bit])

# --- 3. Grover 8-bit Key Recovery Construction ---

# Qubits Register Mapping:
# [0..7]: Key Register (8 qubits)
# [8..15]: State Register (Plaintext XOR Key) (8 qubits)
# [16..23]: Output Ciphertext Register (8 qubits)
# [24..59]: S-box Ancilla Register (36 qubits)
# [60]: Grover Phase Kickback target qubit (1 qubit)

def build_grover_8bit_oracle(pt_val, ct_val):
    oracle_circuit = QuantumCircuit(61)

    # 1. Initialize State Register (8..15) with plaintext
    for i in range(8):
        if (pt_val >> i) & 1:
            oracle_circuit.x(8 + i)

    # 2. XOR Key into State Register
    for i in range(8):
        oracle_circuit.cx(i, 8 + i)

    # 3. Apply 8-bit S-box (Inversion + Affine)
    gf8_inverse(oracle_circuit, list(range(8, 16)), list(range(16, 24)), list(range(24, 60)))
    aes_affine_transformation(oracle_circuit, list(range(16, 24)))

    # 4. Compare with target Ciphertext
    for i in range(8):
        if not ((ct_val >> i) & 1):
            oracle_circuit.x(16 + i)

    # Multi-controlled target on qubit 60
    oracle_circuit.mcx(list(range(16, 24)), 60)

    # 5. Uncompute Comparison
    for i in range(8):
        if not ((ct_val >> i) & 1):
            oracle_circuit.x(16 + i)

    # Uncompute S-box (Self-inverse S-box inverse path)
    aes_affine_transformation(oracle_circuit, list(range(16, 24)))
    gf8_inverse(oracle_circuit, list(range(8, 16)), list(range(16, 24)), list(range(24, 60)))

    # Uncompute Key addition & plaintext setup
    for i in range(8):
        oracle_circuit.cx(i, 8 + i)
    for i in range(8):
        if (pt_val >> i) & 1:
            oracle_circuit.x(8 + i)

    return oracle_circuit

def build_diffusion_operator_8bit():
    diffusion = QuantumCircuit(8)
    diffusion.h(range(8))
    diffusion.x(range(8))
    diffusion.h(7)
    diffusion.mcx(list(range(7)), 7)
    diffusion.h(7)
    diffusion.x(range(8))
    diffusion.h(range(8))
    return diffusion

# Construct the 8-bit Grover search instance
plaintext_8bit = 0x5A
target_ciphertext_8bit = 0x3F

oracle_8bit = build_grover_8bit_oracle(plaintext_8bit, target_ciphertext_8bit)
diffusion_8bit = build_diffusion_operator_8bit()

grover_8bit_circuit = QuantumCircuit(61)
grover_8bit_circuit.h(range(8))
grover_8bit_circuit.x(60)
grover_8bit_circuit.h(60)

# Apply 1 iteration to verify resources
grover_8bit_circuit.compose(oracle_8bit, inplace=True)
grover_8bit_circuit.compose(diffusion_8bit, qubits=list(range(8)), inplace=True)

print("Full 8-bit Grover Circuit setup completed!")
print(f"Total Qubits: {grover_8bit_circuit.num_qubits}")
print(f"Circuit Depth (1 iteration): {grover_8bit_circuit.depth()}")
print("Gate Counts:", grover_8bit_circuit.count_ops())
````
